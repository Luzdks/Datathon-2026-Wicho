```python
#analizar cuantas preguntas hace el mismo usuario en un mismo día
import pandas as pd

# convertir datetime correctamente
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# extraer solo fecha
df['date_only'] = df['date'].dt.date

# agrupar conversaciones por usuario por día
conversaciones_por_dia_usuario = (
    df.groupby(['user_id', 'date_only'])
      .size()
      .reset_index(name='chats_por_dia')
)

conversaciones_por_dia_usuario.head()

#mostrar los 10 usuarios con más días de conversaciones
top_10_usuarios_dias = (
    conversaciones_por_dia_usuario.groupby('user_id')
      .size()
      .reset_index(name='dias_conversacion')
      .sort_values(by='dias_conversacion', ascending=False)
      .head(10)
)
plt.figure(figsize=(10, 6))
sns.barplot(x='user_id', y='dias_conversacion', data=top_10_usuarios_dias, palette='magma')
plt.title('Top 10 Usuarios con Más Días de Conversación')
plt.xlabel('User ID')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================
# 1. PREPROCESAMIENTO
# =========================
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date_only'] = df['date'].dt.date
df['month_year'] = df['date'].dt.to_period('M').astype(str)

# costos
costo_bot = 0.1      # MXN por chat
costo_humano = 30    # MXN por chat

# =========================
# 2. PROMPTS POR USUARIO POR DÍA
# =========================
conversaciones_por_dia_usuario = (
    df.groupby(['user_id', 'date_only'])
      .size()
      .reset_index(name='chats_por_dia')
)

print(conversaciones_por_dia_usuario.head())

# Top 10 usuarios con más días de conversación
top_10_usuarios_dias = (
    conversaciones_por_dia_usuario.groupby('user_id')
      .size()
      .reset_index(name='dias_conversacion')
      .sort_values(by='dias_conversacion', ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(x='user_id', y='dias_conversacion', data=top_10_usuarios_dias, palette='magma')
plt.title('Top 10 Usuarios con Más Días de Conversación')
plt.xlabel('User ID')
plt.ylabel('Días con conversación')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================
# 3. PROMPTS POR USUARIO POR MES
# =========================
conversaciones_por_mes_usuario = (
    df.groupby(['user_id', 'month_year'])
      .size()
      .reset_index(name='chats_por_mes')
)

# Top 10 usuarios con más meses activos
top_10_usuarios_mes = (
    conversaciones_por_mes_usuario.groupby('user_id')
      .size()
      .reset_index(name='meses_conversacion')
      .sort_values(by='meses_conversacion', ascending=False)
      .head(10)
)

print(top_10_usuarios_mes)

# =========================
# 4. PROMPTS POR DÍA
# =========================
conversaciones_por_dia = (
    df.groupby('date_only')
      .size()
      .reset_index(name='total_chats')
)

#CAMBIAR COLOR DE FONDO DE LA GRAFICA, FONDO NEGRO Y LA LIONEA ROSA
sns.set_style("whitegrid")

plt.figure(figsize=(12, 6))
ax = plt.gca()

# Fondo negro
ax.set_facecolor("black")

# Línea rosa
sns.lineplot(
    x='date_only',
    y='total_chats',
    data=conversaciones_por_dia,
    marker='o',
    color='hotpink'
)

# Títulos y etiquetas en blanco
plt.title('Número de Prompts por Día', color='white')
plt.xlabel('Fecha', color='white')
plt.ylabel('Número de Prompts', color='white')

# Ajustar color de ticks
plt.xticks(rotation=45, color='white')
plt.yticks(color='white')

# Colorear los ejes (spines)
for spine in ax.spines.values():
    spine.set_color('white')

# Colorear las líneas de la cuadrícula
ax.grid(color='gray', linestyle='--', linewidth=0.5)

plt.tight_layout()
plt.show()


# =========================
# 5. COSTO Y AHORRO POR DÍA
# =========================
conversaciones_por_dia['costo_bot_dia'] = conversaciones_por_dia['total_chats'] * costo_bot
conversaciones_por_dia['costo_humano_dia'] = conversaciones_por_dia['total_chats'] * costo_humano
conversaciones_por_dia['ahorro_dia'] = conversaciones_por_dia['costo_humano_dia'] - conversaciones_por_dia['costo_bot_dia']

print(
    conversaciones_por_dia
    .sort_values(by='total_chats', ascending=False)
    [['date_only', 'total_chats', 'costo_bot_dia', 'costo_humano_dia', 'ahorro_dia']]
    .head(10)
)

# =========================
# 6. AHORRO POR MES
# =========================
conversaciones_por_mes = (
    df.groupby('month_year')
      .size()
      .reset_index(name='total_chats')
)

conversaciones_por_mes['costo_bot_mes'] = conversaciones_por_mes['total_chats'] * costo_bot
conversaciones_por_mes['costo_humano_mes'] = conversaciones_por_mes['total_chats'] * costo_humano
conversaciones_por_mes['ahorro_mes'] = conversaciones_por_mes['costo_humano_mes'] - conversaciones_por_mes['costo_bot_mes']

plt.figure(figsize=(12, 6))
sns.lineplot(x='month_year', y='ahorro_mes', data=conversaciones_por_mes, marker='o')
plt.title('Ahorro Mensual al Usar el Bot en Lugar de Humanos')
plt.xlabel('Mes')
plt.ylabel('Ahorro estimado (MXN)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(conversaciones_por_mes)
```
