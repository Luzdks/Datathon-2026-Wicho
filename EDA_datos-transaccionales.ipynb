```python
df_clientes = pd.read_csv("data/dataset_transacciones/hey_clientes.csv")
df_productos = pd.read_csv("data/dataset_transacciones/hey_productos.csv")
df_trans = pd.read_csv("data/dataset_transacciones/hey_transacciones.csv")
```

```python
df_clientes_limpio = df_clientes.dropna(subset=['estado'])
df_clientes_limpio.isnull().sum()
```

```python
df_clientes_limpio = df_clientes_limpio.dropna(subset=['satisfaccion_1_10'])
df_clientes_limpio.isnull().sum()
```

```python
df_clientes_limpio.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df_clientes_limpio.select_dtypes(include=['float64', 'int64']).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt = '.2f', square=True, linewidths=0.5)
plt.title('Correlation Heatmap of Numeric Features', fontsize=14)
plt.show()
```

```python
numeric_df = df_clientes_limpio.select_dtypes(include=['float64', 'int64'])
sns.pairplot(numeric_df, diag_kind='kde', plot_kws={'alpha': 0.6, 's': 40, 'edgecolor': 'k'})
plt.suptitle('Scatter Matrix of Numerical Features', fontsize=16, y=1.02)
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

X = df_clientes_limpio[['edad', 'ingreso_mensual_mxn', 'satisfaccion_1_10', 'antiguedad_dias', 'score_buro', 'dias_desde_ultimo_login', 'num_productos_activos']]

X = add_constant(X)

vif = pd.DataFrame()
vif['Feature'] = X.columns
vif['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

vif
```

```python
# Distribución de estados
plt.figure(figsize=(10, 5))
df_clientes['estado'].value_counts().plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Distribución de Estados', fontsize=12, fontweight='bold')
plt.xlabel('Estado')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Distribución de usuarios Hey Pro
plt.figure(figsize=(10, 5))
df_clientes['es_hey_pro'].value_counts().plot(kind='bar', color=['#FF6B6B', '#4ECDC4'], edgecolor='black')
plt.title('Distribución de Usuarios: Hey Pro vs No Hey Pro', fontsize=12, fontweight='bold')
plt.xlabel('Hey Pro')
plt.ylabel('Cantidad de Usuarios')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Tabla resumen
print("\nResumen de usuarios Hey Pro:")
print(df_clientes['es_hey_pro'].value_counts())
print(f"\nPorcentaje:")
print(df_clientes['es_hey_pro'].value_counts(normalize=True) * 100)
```
